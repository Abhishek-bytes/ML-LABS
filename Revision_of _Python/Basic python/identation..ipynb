```python
revenues=[20,40,60,80,100]
expenses=[5,15,25,35,55]
revenues,expenses
```

```python
for rev in revenues:
    print(rev)   
```

```python
for rev in revenues:
    print ("revenues:",rev)
```

```python
for rev in revenues:
    print(f"revenues:{rev}") #python formatting 
```

```python
for i in range(len(revenues)):
    profit = revenues[i]-expenses[i]
    margin = profit *100/revenues[i]
    print (f"margin {i +1}: {margin:.2f}" )   # :.2F is gives decimal upto two positions  #i+1 because index starts from 1 
```

```python
margins =[]
for i in range(len(revenues)):
    profit = revenues[i]-expenses[i]
    margin = profit *100/revenues[i]
    margins.append(margin)
margins
```
