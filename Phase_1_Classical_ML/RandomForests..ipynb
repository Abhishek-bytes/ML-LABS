```python
inputs = df.drop('salary_more_then_100k',axis= 'columns')
inputs
```

```python
target =df['salary_more_then_100k']
target
```

Convert the text columns into numeric columns before training. Scikit-learn models cannot fit directly on strings.

```python
inputs_encoded = pd.get_dummies(inputs, columns=['company', 'job', 'degree'], dtype=int)
inputs_encoded
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
model = RandomForestClassifier(random_state=42)
model.fit(inputs_encoded, target)
```

```python
model.score(inputs_encoded, target)
```
